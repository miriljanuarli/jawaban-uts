```python
import pandas as pd

# Load dataset
df = pd.read_csv('/content/winequality-red.csv')

# Membagi data menjadi high dan low berdasarkan median
for col in ['alcohol', 'pH', 'residual sugar', 'citric acid']:
    median_value = df[col].median()

    df[col+'_category'] = df[col].apply(
        lambda x: 'high' if x >= median_value else 'low'
    )

    result = df.groupby(col+'_category')['quality'].mean()

    print("\n", col)
    print(result)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
df = pd.read_csv('/content/winequality-red.csv')

# Informasi data
print(df.info())

# Statistik deskriptif
print(df.describe())

# Cek missing value
print(df.isnull().sum())

# Histogram
df.hist(figsize=(15,10))
plt.show()

# Heatmap korelasi
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.show()

# Boxplot kualitas wine
plt.figure(figsize=(8,5))
sns.boxplot(x='quality', y='alcohol', data=df)
plt.show()
```
